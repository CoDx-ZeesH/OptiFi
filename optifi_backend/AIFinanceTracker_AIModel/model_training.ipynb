```python
import re
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
```

```python
def clean_text(text):
    if isinstance(text,str):
        text=re.sub(r'\d+','',text)
        text=re.sub(r'[^\w\s]','',text)
        words=text.split()
        sw=set(stopwords.words('english'))
        words=[word for word in words if word not in sw]
        return ' '.join(words)
    else:
        return text
```

```python
from sklearn.model_selection import train_test_split
import numpy as np
# Ensure target variable is a 1-D numpy array named `y` to avoid shape mismatches
y = df['category'].values if 'y' not in globals() else np.asarray(y).ravel()
# Split the features and target consistently (use X_final which is the combined text+numeric features)
X_train, X_test, Y_train, Y_test = train_test_split(X_final, y, test_size=0.2, random_state=42)
print('X_train shape:', getattr(X_train, 'shape', None))
print('Y_train shape:', getattr(Y_train, 'shape', None))
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report,confusion_matrix,accuracy_score

model = LogisticRegression(max_iter=5000, solver='lbfgs')
model.fit(X_train,Y_train)
Y_pred=model.predict(X_test)

print("Accuracy :",accuracy_score(Y_test,Y_pred))
```

```python
print(classification_report(Y_test, Y_pred))
print(confusion_matrix(Y_test, Y_pred))
```
